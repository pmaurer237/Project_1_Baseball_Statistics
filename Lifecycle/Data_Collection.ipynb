## FanGraphs Stats Collection (end of demonstration)

### Necessary Packages

```python
import pandas as pd 
from pybaseball import playerid_reverse_lookup
from pybaseball import statcast_batter_exitvelo_barrels
```

### Setup

```python
# Set some constants
START_SEASON = 2017
END_SEASON = 2024
MIN_PA = 200  # Set the minimum number of season plate appearances to include a player
SEASONS = list(range(START_SEASON, END_SEASON + 1))
```

### Pull (2017-2024)

```python
# Pull FanGraphs stats for all MLB hitters from 2017 to 2024, excluding 2020
fangraphs_list = []
for season in SEASONS:
    if season == 2020:
        continue  # Skip 2020 COVID season
    df_f = batting_stats(season, qual = 0)  
    
    # Filter for hitters with at least 200 seasonal plate appearances
    df_f = df_f[df_f["PA"] >= MIN_PA]
    fangraphs_list.append(df_f)

# Concatenate all seasons 
fangraphs_raw = pd.concat(fangraphs_list, ignore_index = True)
print(f"Shape of FanGraphs dataset for all players initially: {fangraphs_raw.shape}\n")
fangraphs_raw.head()
```

### FanGraphs IDs

```python
# Fix the FanGraphs ID column name 
fangraphs_raw = fangraphs_raw.rename(columns = {"IDfg": "fangraphs_id"})

# Ensure that FanGraphs IDs are integers
fangraphs_raw["fangraphs_id"] = pd.to_numeric(fangraphs_raw["fangraphs_id"], 
                                              errors = "coerce").astype("Int64")

# Check for missing FanGraphs IDs
missing_fg = fangraphs_raw["fangraphs_id"].isna().sum()
print(f"Missing FanGraphs IDs: {missing_fg}\n")

# Display duplicates by name and season
duplicates = fangraphs_raw[fangraphs_raw.duplicated(subset = ["Name", "Season"], 
                                                    keep = False)]
print(f"Number of duplicate name-season pairs: {duplicates.shape[0]}")
```

## Statcast Data Collection

### ID Map: FanGraphs $\longrightarrow$ MLBAM (for Statcast)

```python
# Collect unique FanGraphs IDs from the FanGraphs dataset
unique_fangraphs_ids = fangraphs_raw["fangraphs_id"].dropna().astype("Int64").unique().tolist()

# Look up cross-IDs for all of these FanGraphs IDs
id_map = playerid_reverse_lookup(unique_fangraphs_ids, key_type = "fangraphs")

# Keep the relevant IDs and rename the columns
id_map = id_map.rename(columns = {"key_fangraphs": "fangraphs_id",
                                  "key_mlbam": "mlbam_id"})
id_map = id_map[["fangraphs_id", "mlbam_id"]]
print("IDs:\n")
id_map.head()
```

### New MLBAM ID Column, FanGraphs Dataset Merge

```python
# Left-join the MLBAM ID column to the FanGraphs dataset
fangraphs_raw = fangraphs_raw.merge(id_map, on = "fangraphs_id", how = "left")

# Ensure that MLBAM IDs are integers
fangraphs_raw["mlbam_id"] = pd.to_numeric(fangraphs_raw["mlbam_id"], 
                                          errors = "coerce").astype("Int64")

# Check for missing MLBAM IDs
missing_mlbam = fangraphs_raw["mlbam_id"].isna().sum()
print(f"Missing MLBAM IDs: {missing_mlbam}\n")
fangraphs_raw.head()
```

### Players with No MLBAM ID

```python
# Remove players with no MLBAM ID since they have no Statcast data
before = fangraphs_raw.shape[0]
fangraphs_raw = fangraphs_raw.dropna(subset = ["mlbam_id"])
fangraphs_raw["mlbam_id"] = fangraphs_raw["mlbam_id"].astype("Int64")
after = fangraphs_raw.shape[0]

# Report how many players were removed
print(f"Removed {before - after} players with missing MLBAM IDs!\n")
print(f"Remaining rows: {after}\n")
fangraphs_raw.head()
```

### NOW FanGraphs Dataset Save

```python
# Save the dataset to a CSV file
fangraphs_raw.to_csv("../Data/Raw/FanGraphs_Raw.csv", index = False)
```

### Pull (2017-2024)

```python
# Pull Statcast aggregates for all hitters from 2017 to 2024, excluding 2020
statcast_list = []
for season in SEASONS:
    if season == 2020:  
        continue  # Skip 2020 COVID season
    df_s = statcast_batter_exitvelo_barrels(season)
    df_s["Season"] = season
    statcast_list.append(df_s)

# Concatenate all seasons 
statcast_raw = pd.concat(statcast_list, ignore_index = True)
print(f"Shape of Statcast dataset for all players initially: {statcast_raw.shape}\n")
statcast_raw.head()
```

Note: The Statcast batting dataset is smaller (in terms of rows) than the FanGraphs batting dataset because Statcast only includes players who generated measurable batted-ball events in a given season. While FanGraphs reports statistics for every hitter who recorded plate appearances, Statcast requires actual tracked contact, such as exit velocity or launch angle, to generate a profile. As a result, players with limited playing time, few balls in play, injury-shortened seasons, or data inconsistencies may be present in FanGraphs but missing from Statcast. This discrepancy is expected and does not affect the analysis, since all meaningful comparison hitters, including Aaron Judge, have complete Statcast data.

### Row Removal, Statcast Dataset Save

```python
# Keep Statcast rows only for hitters who appear in the FanGraphs dataset
valid_mlbam_ids = fangraphs_raw["mlbam_id"].unique().tolist()
statcast_raw = statcast_raw[statcast_raw["player_id"].isin(valid_mlbam_ids)]
print(f"Statcast rows after filtering to hitters with FanGraphs data: {statcast_raw.shape[0]}\n")

# Save the dataset to a CSV file
statcast_raw.to_csv("../Data/Raw/Statcast_Raw.csv", index = False)
```
