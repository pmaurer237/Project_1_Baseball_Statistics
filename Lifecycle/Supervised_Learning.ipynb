## Univariate Linear Regression

Let $l\colon\mathbb{R}\to\mathbb{R}$ be defined as

$$
\hat{y}_n=l(x_n)=\theta_0+\theta_1x_n \tag*{(1)}
$$

which is a proposed univariate linear model for the dataset $D=\left\{(x_1,\,y_1),\,(x_2,\,y_2),\,\ldots,\,(x_N,\,y_N)\right\}$, where $\theta_0,\,\theta_1$ are unknown parameters.

Then, the residuals are 

$$
\varepsilon_n=y_n-\hat{y}_n \tag*{(2)}
$$

and the error or loss function in this case is the mean squared error (MSE)

$$
\mathscr{L}(\theta_0,\,\theta_1)=\frac{1}{N}\sum_{n=1}^N\varepsilon_n^2=\frac{1}{N}\sum_{n=1}^N(y_n-\hat{y}_n)^2=\frac{1}{N}\sum_{n=1}^N(y_n-\theta_0-\theta_1x_n)^2 \tag*{(3)}
$$

Minimizing the MSE yields closed-form expressions for the slope and intercept, which are referred to as the ordinary least squares (OLS) estimates of the parameters $\theta_0$ and $\theta_1$:

$$
\hat{\theta}_0=\overline{y}-\hat{\theta}_1\overline{x},\qquad\hat{\theta}_1=\frac{\sum_{n=1}^N(x_n-\overline{x})(y_n-\overline{y})}{\sum_{n=1}^N(x_n-\overline{x})^2} \tag*{(4)}
$$

As a reminder, 

$$
\overline{x}=\frac{1}{N}\sum_{n=1}^Nx_n,\qquad\overline{y}=\frac{1}{N}\sum_{n=1}^Ny_n \tag*{(5)}
$$

## Multivariate Logistic Classification

Let $\sigma\colon\mathbb{R}\to(0,\,1)$ be the sigmoid (logistic) function defined as

$$
\sigma(z)=\frac{1}{1+e^{-z}} \tag*{(8)}
$$

A useful property of $\sigma$ is that its derivative can be written in terms of $\sigma$ itself:

$$
\sigma'(z)=\frac{e^{-z}}{\left(1+e^{-z}\right)^2}=\sigma(z)\big(1-\sigma(z)\big) \tag*{(9)}
$$

For a $k$-dimensional feature vector $\vec{x}=(x_1,\,x_2,\,\ldots,\,x_k)\in\mathbb{R}^{1\times k}$, the augmented feature vector is $\vec{x}^*=(1,\,x_1,\,x_2,\,\ldots,\,x_k)\in\mathbb{R}^{1\times(k+1)}$. 

Let the parameter vector be $\vec{w}=(w_0,\,w_1,\,\ldots,\,w_k)^\top\in\mathbb{R}^{(k+1)\times1}$, so that

$$
z=\vec{x}^*\,\vec{w}=w_0+w_1x_1+w_2x_2+\cdots+w_kx_k \tag*{(10)}
$$

is a scalar.

Hence, the multivariate logistic classifier predicts

$$
\hat{y}=\sigma(z)=P(y=1\mid\vec{x}) \tag*{(11)}
$$

where $\hat{y}\in(0,\,1)$ is interpreted as the probability of the positive class $y=1$, while the negative class is $y=0$.

To turn this probability into a class label $\hat{c}\in\{0,\,1\}$, a threshold $\tau\in(0,\,1)$ is chosen:

$$
\hat{c}=
\begin{cases}
    1, & \hat{y}\geq\tau \\
    0, & \hat{y}<\tau
\end{cases}
\tag*{(12)}
$$

For the default $\tau=0.5$, the condition $\hat{y}\geq0.5$ is equivalent to $z=\vec{x}^*\,\vec{w}\geq0$, so the decision boundary is a hyperplane in feature space. In other words, logistic classification is a linear classifier, even though $\sigma$ is nonlinear.

Since $y$ is a Bernoulli random variable, $P(y\mid\vec{x})=\hat{y}^{\,y}(1-\hat{y})^{1-y}$, and taking the negative natural logarithm gives the loss for a single observation, the binary cross-entropy (BCE):

$$
\ell(y,\,\hat{y})=-\big(y\ln(\hat{y})+(1-y)\ln(1-\hat{y})\big) \tag*{(13)}
$$

Differentiating with respect to $\hat{y}$ gives

$$
\frac{\partial\ell}{\partial\hat{y}}=-\frac{y}{\hat{y}}+\frac{1-y}{1-\hat{y}}=\frac{\hat{y}-y}{\hat{y}(1-\hat{y})} \tag*{(14)}
$$

Then, since $\hat{y}=\sigma(z)$, the chain rule and $(8)$ give

$$
\frac{\partial\ell}{\partial z}=\frac{\partial\ell}{\partial\hat{y}}\cdot\frac{d\hat{y}}{dz}=\frac{\hat{y}-y}{\hat{y}(1-\hat{y})}\cdot\hat{y}(1-\hat{y})=\hat{y}-y \tag*{(15)}
$$

Furthermore, from $(9)$, $\nabla_{\vec{w}}\,z=\vec{x}^{*\top}$, so the gradient of the single-observation loss with respect to the parameter vector is

$$
\nabla_{\vec{w}}\,\ell=(\hat{y}-y)\,\vec{x}^{*\top}\in\mathbb{R}^{(k+1)\times1} \tag*{(16)}
$$

For a dataset with $N$ observations $\left\{(\vec{x}_1,\,y_1),\,(\vec{x}_2,\,y_2),\,\ldots,\,(\vec{x}_N,\,y_N)\right\}$, the objective function is the mean BCE:

$$
\mathscr{L}(\vec{w})=\frac{1}{N}\sum_{n=1}^N\ell\left(y_n,\,\sigma\left(\vec{x}_n^*\,\vec{w}\right)\right) \tag*{(17)}
$$

Thus, the gradient of the objective function is

$$
\nabla\mathscr{L}(\vec{w})=\frac{1}{N}\sum_{n=1}^N(\hat{y}_n-y_n)\,\vec{x}_n^{*\top}=\frac{1}{N}\sum_{n=1}^N\left(\sigma\left(\vec{x}_n^*\,\vec{w}\right)-y_n\right)\vec{x}_n^{*\top} \tag*{(18)}
$$

Equivalently, if $D\in\mathbb{R}^{N\times(k+1)}$ is the design matrix whose $n\text{th}$ row is $\vec{x}_n^*$, $\vec{y}=(y_1,\,y_2,\,\ldots,\,y_N)^\top$, and $\sigma$ is applied element-wise, then

$$
\nabla\mathscr{L}(\vec{w})=\frac{1}{N}\,D^\top\left(\sigma\left(D\vec{w}\right)-\vec{y}\right) \tag*{(19)}
$$

Unlike linear regression, setting $\nabla\mathscr{L}(\vec{w})=\vec{0}$ has no closed-form solution for the optimal parameter vector, so iterative optimization methods such as gradient descent are used to minimize $\mathscr{L}$. 

Fortunately, $\mathscr{L}$ is convex in $\vec{w}$, so any local minimum is a global minimum.

Hence, the update rule is

$$
\vec{w}^{(t+1)}=\vec{w}^{(t)}-\xi\,\nabla\mathscr{L}\left(\vec{w}^{(t)}\right) \tag*{(20)}
$$

$$
=\vec{w}^{(t)}-\frac{\xi}{N}\sum_{n=1}^N\left(\sigma\left(\vec{x}_n^*\,\vec{w}^{(t)}\right)-y_n\right)\vec{x}_n^{*\top} \tag*{(21)}
$$

where $\xi\in\mathbb{R}^+$ is the learning rate and $t$ is the iteration of the parameter vector.

Note: If the two classes are perfectly linearly separable, $\mathscr{L}$ has no finite minimizer, and $\|\vec{w}\|$ grows without bound as gradient descent continues. For this reason (and to reduce overfitting), a penalty term is usually added to $\mathscr{L}$. In fact, `sklearn`'s `LogisticRegression` applies an $L_2$ penalty on $(w_1,\,\ldots,\,w_k)$ by default, with its strength controlled by the hyperparameter `C` (smaller `C` means a stronger penalty), which is a natural candidate for `GridSearchCV`.
