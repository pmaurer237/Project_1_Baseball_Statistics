## Univariate Linear Regression

Let $l\colon\mathbb{R}\to\mathbb{R}$ be defined as

$$
\hat{y}_i=l(x_i)=\theta_0+\theta_1x_i \tag*{(1)}
$$

which is a proposed univariate linear model for the dataset $D=\left\{(x_1,\,y_1),\,(x_2,\,y_2),\,\ldots,\,(x_N,\,y_N)\right\}$, where $\theta_0,\,\theta_1$ are unknown parameters.

Then, the residuals are defined as

$$
\varepsilon_i=y_i-\hat{y}_i \tag*{(2)}
$$

and the error or loss function in this case can be defined as the mean squared error (MSE)

$$
\ell(\theta_0,\,\theta_1)=\frac{1}{N}\sum_{i=1}^N\varepsilon_i^2=\frac{1}{N}\sum_{i=1}^N(y_i-\hat{y}_i)^2=\frac{1}{N}\sum_{i=1}^N(y_i-\theta_0-\theta_1x_i)^2 \tag*{(3)}
$$

Minimizing the MSE yields closed-form expressions for the slope and intercept. These are called the ordinary least squares (OLS) estimates of the parameters $\theta_0$ and $\theta_1$:

$$
\hat{\theta}_0=\overline{y}-\hat{\theta}_1\overline{x},\qquad\hat{\theta}_1=\frac{\sum_{i=1}^N(x_i-\overline{x})(y_i-\overline{y})}{\sum_{i=1}^N(x_i-\overline{x})^2} \tag*{(4)}
$$

As a reminder, 

$$
\overline{x}=\frac{1}{N}\sum_{i=1}^Nx_i,\qquad\overline{y}=\frac{1}{N}\sum_{i=1}^Ny_i \tag*{(5)}
$$

## Multivariate Logistic Classification

Let $\sigma\colon\mathbb{R}\to\mathbb{R}$ be defined as

$$
\sigma(z)=\frac{1}{1+e^{-z}} \tag*{(6)}
$$

For a $k$-dimensional feature vector $\vec{x}=(x_1,\,x_2,\,\ldots,\,x_k)$, define the augmented feature vector as

$$
\vec{x}^*=(1,\,x_1,\,x_2,\,\ldots,\,x_k)
$$

and let the parameter vector satisfy $\vec{w}\in\mathbb{R}^{k+1}$.

Now define

$$
z=\vec{w}^{\top}\cdot\vec{x}^*=w_0+w_1x_1+w_2x_2+\cdots+w_kx_k \tag*{(7)}
$$

Hence, the multivariate logistic classifier predicts

$$
\hat{y}=\sigma(z)
$$

where $\hat{y}\in(0,\,1)$ is interpreted as the probability of the positive class $(y=1)$, while the negative class is $y=0$.

The error function here is defined as the binary cross-entropy (BCE):

$$
\beta(y,\,\hat{y})=-\big(y\ln(\hat{y})+(1-y)\ln(1-\hat{y})\big) \tag*{(8)}
$$

Since $\hat{y}=\sigma(z)$, differentiation with respect to $z$ gives

$$
\frac{d\beta}{dz}=\hat{y}-y
$$

Therefore, by the chain rule, the gradient with respect to the parameter vector is

$$
\nabla\beta(\vec{w})=(\hat{y}-y)\vec{x}^* \tag*{(9)}
$$

Thus, the total mean gradient for a dataset with $N$ observations is defined as

$$
\nabla J(\vec{w})=\frac{1}{N}\sum_{i=1}^N(\hat{y}_i-y_i)\vec{x}_i^*=\frac{1}{N}\sum_{i=1}^N\left(\sigma(\vec{w}^{\top}\cdot\vec{x}_i^*)-y_i\right)\vec{x}_i^* \tag*{(10)}
$$

Unlike linear regression, there is no closed-form solution for the optimal parameter vector, so iterative optimization methods such as gradient descent are used.

Hence, the update rule is

$$
\vec{w}_{j+1}=\vec{w}_j-\xi\,\nabla J(\vec{w}_j) \tag*{(11)}
$$

$$
=\vec{w}_j-\xi\sum_{i=1}^N\left(\sigma(\vec{w}_j^{\top}\cdot\vec{x}_i^*)-y_i\right)\vec{x}_i^* \tag*{(12)}
$$

where $\xi\in\mathbb{R}^+$ is the step size or learning rate and $j$ is the iteration of the parameter vector.
