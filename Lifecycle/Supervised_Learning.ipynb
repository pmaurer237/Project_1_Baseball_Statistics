## Simple Linear Regression

Let $l\colon\mathbb{R}\to\mathbb{R}$ be defined as

$$
\hat{y}_n=l(x_n)=\theta_0+\theta_1x_n \tag*{(1)}
$$

which is a proposed simple linear model for the dataset $\left\{(x_1,\,y_1),\,(x_2,\,y_2),\,\ldots,\,(x_N,\,y_N)\right\}$, where $\theta_0,\,\theta_1$ are unknown parameters.

Then, the residuals are defined as

$$
\varepsilon_n=y_n-\hat{y}_n \tag*{(2)}
$$

and the error or loss function, in this case, is defined as the mean squared error (MSE):

$$
\mathscr{L}(\theta_0,\,\theta_1)=\frac{1}{N}\sum_{n=1}^N\varepsilon_n^2=\frac{1}{N}\sum_{n=1}^N(y_n-\hat{y}_n)^2=\frac{1}{N}\sum_{n=1}^N(y_n-\theta_0-\theta_1x_n)^2 \tag*{(3)}
$$

Minimizing the MSE yields closed-form solutions for the slope and intercept, which are referred to as the ordinary least squares (OLS) estimates of the parameters $\theta_0$ and $\theta_1$. Since $\mathscr{L}$ is differentiable in both parameters, its minimum occurs where both partial derivatives equal zero:

$$
\frac{\partial}{\partial\theta_0}\left(\frac{1}{N}\sum_{n=1}^N(y_n-\theta_0-\theta_1x_n)^2\right)=0,\qquad\frac{\partial}{\partial\theta_1}\left(\frac{1}{N}\sum_{n=1}^N(y_n-\theta_0-\theta_1x_n)^2\right)=0
$$

$$
\frac{1}{N}\sum_{n=1}^N\frac{\partial}{\partial\theta_0}(y_n-\theta_0-\theta_1x_n)^2=0,\qquad\frac{1}{N}\sum_{n=1}^N\frac{\partial}{\partial\theta_1}(y_n-\theta_0-\theta_1x_n)^2=0
$$

Applying the chain rule to each term gives

$$
-\frac{2}{N}\sum_{n=1}^N(y_n-\theta_0-\theta_1x_n)=0,\qquad-\frac{2}{N}\sum_{n=1}^Nx_n(y_n-\theta_0-\theta_1x_n)=0
$$

and multiplying both equations by $-N/2$ yields the normal equations:

$$
\sum_{n=1}^N(y_n-\theta_0-\theta_1x_n)=0,\qquad\sum_{n=1}^Nx_n(y_n-\theta_0-\theta_1x_n)=0
$$

Let the sample means be

$$
\overline{x}=\frac{1}{N}\sum_{n=1}^Nx_n,\qquad\overline{y}=\frac{1}{N}\sum_{n=1}^Ny_n \tag*{(4)}
$$

so that $\sum_{n=1}^Nx_n=N\overline{x}$ and $\sum_{n=1}^Ny_n=N\overline{y}$. Distributing the sum in the first normal equation then gives

$$
N\overline{y}-N\theta_0-\theta_1N\overline{x}=0\implies\theta_0=\overline{y}-\theta_1\overline{x}
$$

Substituting this expression for $\theta_0$ into the second normal equation gives

$$
\sum_{n=1}^Nx_n\big((y_n-\overline{y})-\theta_1(x_n-\overline{x})\big)=0\implies\theta_1=\frac{\sum_{n=1}^Nx_n(y_n-\overline{y})}{\sum_{n=1}^Nx_n(x_n-\overline{x})}
$$

Since deviations from a mean sum to zero, $\sum_{n=1}^N(x_n-\overline{x})=0$ and $\sum_{n=1}^N(y_n-\overline{y})=0$. Hence, $\overline{x}\sum_{n=1}^N(y_n-\overline{y})=0$ and $\overline{x}\sum_{n=1}^N(x_n-\overline{x})=0$, and subtracting these zeros from the numerator and denominator, respectively, replaces $x_n$ with $x_n-\overline{x}$:

$$
\sum_{n=1}^Nx_n(y_n-\overline{y})=\sum_{n=1}^N(x_n-\overline{x})(y_n-\overline{y}),\qquad\sum_{n=1}^Nx_n(x_n-\overline{x})=\sum_{n=1}^N(x_n-\overline{x})^2
$$

Thus, the OLS estimates are

$$
\hat{\theta}_0=\overline{y}-\hat{\theta}_1\overline{x},\qquad\hat{\theta}_1=\frac{\sum_{n=1}^N(x_n-\overline{x})(y_n-\overline{y})}{\sum_{n=1}^N(x_n-\overline{x})^2} \tag*{(5)}
$$

Note: $\hat{\theta}_1$ is defined only when $\sum_{n=1}^N(x_n-\overline{x})^2>0$, that is, when the $x_n$ are not all equal. Furthermore, $\mathscr{L}$ is a sum of squares of functions that are linear in $\theta_0$ and $\theta_1$, so it is convex, and this critical point is therefore the global minimum rather than a maximum or saddle point.

## Multiple Logistic Regression

Let $\sigma\colon\mathbb{R}\to(0,\,1)$ be the sigmoid (logistic) function defined as

$$
\sigma(z)=\frac{1}{1+e^{-z}} \tag*{(6)}
$$

A useful property of $\sigma$ is that its derivative can be written in terms of $\sigma$ itself:

$$
\sigma'(z)=\frac{e^{-z}}{\left(1+e^{-z}\right)^2}=\sigma(z)\big(1-\sigma(z)\big) \tag*{(7)}
$$

For a $k$-dimensional feature vector $\vec{x}=(x_1,\,x_2,\,\ldots,\,x_k)\in\mathbb{R}^{1\times k}$, the augmented feature vector is $\vec{x}^*=(1,\,x_1,\,x_2,\,\ldots,\,x_k)\in\mathbb{R}^{1\times(k+1)}$. 

Let the parameter vector be $\vec{w}=(w_0,\,w_1,\,\ldots,\,w_k)^\top\in\mathbb{R}^{(k+1)\times1}$, so that

$$
z=\vec{x}^*\,\vec{w}=w_0+w_1x_1+w_2x_2+\cdots+w_kx_k \tag*{(8)}
$$

is a scalar.

The multiple logistic regression model predicts

$$
\hat{y}=\sigma(z)=P(y=1\mid\vec{x}) \tag*{(9)}
$$

where $\hat{y}\in(0,\,1)$ is interpreted as the probability of the positive class $y=1$, while the negative class is $y=0$.

To turn this probability into a class label $\hat{c}\in\{0,\,1\}$, a threshold $\tau\in(0,\,1)$ is chosen:

$$
\hat{c}=
\begin{cases}
    1, & \hat{y}>\tau \\
    0, & \hat{y}\leq\tau
\end{cases}
\tag*{(10)}
$$

For the default $\tau=0.5$, the condition $\hat{y}>0.5$ is equivalent to $z=\vec{x}^*\,\vec{w}>0$, so the decision boundary $\vec{x}^*\,\vec{w}=0$ is a hyperplane in feature space. In other words, logistic regression is a linear classifier, even though $\sigma$ is nonlinear. The strict inequality matches `sklearn`, whose `predict` method assigns class $1$ only when $z>0$. A point exactly on the boundary is assigned class $0$.

Since $y$ is a Bernoulli random variable, $P(y\mid\vec{x})=\hat{y}^{\,y}(1-\hat{y})^{1-y}$, and taking the negative natural logarithm gives the loss for a single observation, the binary cross-entropy (BCE):

$$
\ell(y,\,\hat{y})=-\big(y\ln(\hat{y})+(1-y)\ln(1-\hat{y})\big) \tag*{(11)}
$$

Differentiating with respect to $\hat{y}$ gives

$$
\frac{\partial\ell}{\partial\hat{y}}=-\frac{y}{\hat{y}}+\frac{1-y}{1-\hat{y}}=\frac{\hat{y}-y}{\hat{y}(1-\hat{y})} \tag*{(12)}
$$

Then, since $\hat{y}=\sigma(z)$, the chain rule and $(7)$ give

$$
\frac{\partial\ell}{\partial z}=\frac{\partial\ell}{\partial\hat{y}}\cdot\frac{d\hat{y}}{dz}=\frac{\hat{y}-y}{\hat{y}(1-\hat{y})}\cdot\hat{y}(1-\hat{y})=\hat{y}-y \tag*{(13)}
$$

Furthermore, from $(8)$, $\nabla_{\vec{w}}\,z=\vec{x}^{*\top}$, so the gradient of the single-observation loss with respect to the parameter vector is

$$
\nabla_{\vec{w}}\,\ell=(\hat{y}-y)\,\vec{x}^{*\top}\in\mathbb{R}^{(k+1)\times1} \tag*{(14)}
$$

For a dataset with $N$ observations $\left\{(\vec{x}_1,\,y_1),\,(\vec{x}_2,\,y_2),\,\ldots,\,(\vec{x}_N,\,y_N)\right\}$, the objective function is the mean BCE:

$$
\mathscr{L}(\vec{w})=\frac{1}{N}\sum_{n=1}^N\ell\left(y_n,\,\sigma\left(\vec{x}_n^*\,\vec{w}\right)\right) \tag*{(15)}
$$

Thus, the gradient of the objective function is

$$
\nabla\mathscr{L}(\vec{w})=\frac{1}{N}\sum_{n=1}^N(\hat{y}_n-y_n)\,\vec{x}_n^{*\top}=\frac{1}{N}\sum_{n=1}^N\left(\sigma\left(\vec{x}_n^*\,\vec{w}\right)-y_n\right)\vec{x}_n^{*\top} \tag*{(16)}
$$

Equivalently, if $X\in\mathbb{R}^{N\times(k+1)}$ is the design matrix whose $n\text{th}$ row is $\vec{x}_n^*$, $\vec{y}=(y_1,\,y_2,\,\ldots,\,y_N)^\top$, and $\sigma$ is applied element-wise, then

$$
\nabla\mathscr{L}(\vec{w})=\frac{1}{N}\,X^\top\left(\sigma\left(X\vec{w}\right)-\vec{y}\right) \tag*{(17)}
$$

Unlike linear regression, setting $\nabla\mathscr{L}(\vec{w})=\vec{0}$ has no closed-form solution for the optimal parameter vector, so iterative optimization methods are used to minimize $\mathscr{L}$. Fortunately, $\mathscr{L}$ is convex in $\vec{w}$, so any local minimum is a global minimum.

The simplest such method is gradient descent. Starting from an initial guess $\vec{w}^{(0)}$, it repeatedly takes a step in the direction of steepest descent, the negative gradient:

$$
\begin{aligned}
\vec{w}^{(t+1)}&=\vec{w}^{(t)}-\xi\,\nabla\mathscr{L}\left(\vec{w}^{(t)}\right) \\
&=\vec{w}^{(t)}-\frac{\xi}{N}\sum_{n=1}^N\left(\sigma\left(\vec{x}_n^*\,\vec{w}^{(t)}\right)-y_n\right)\vec{x}_n^{*\top}
\end{aligned}
\tag*{(18)}
$$

where $\xi\in\mathbb{R}^+$ is the learning rate and $t$ indexes the iteration.

Note: If the two classes are perfectly linearly separable, $\mathscr{L}$ has no finite minimizer, and $\|\vec{w}\|$ grows without bound as gradient descent continues. For this reason (and to reduce overfitting), a penalty term is usually added to $\mathscr{L}$. In fact, `sklearn`'s `LogisticRegression` applies an $L_2$ penalty on $(w_1,\,\ldots,\,w_k)$ by default, with its strength controlled by the hyperparameter `C` (smaller `C` means a stronger penalty), which is a natural candidate for `GridSearchCV`.

In practice, `sklearn`'s `LogisticRegression` also does not use plain gradient descent. Its default solver, `lbfgs` (limited-memory BFGS), is a quasi-Newton method: it uses the gradient of the penalized objective, which is $(16)$ plus the gradient of the penalty term, but it also builds an approximation of the curvature of the objective from recent gradients, which typically lets it converge in far fewer iterations.
