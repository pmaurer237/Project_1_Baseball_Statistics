## Univariate Linear Regression

Let $l\colon\mathbb{R}\to\mathbb{R}$ be defined as

$$
\hat{y}_n=l(x_n)=\theta_0+\theta_1x_n \tag*{(1)}
$$

which is a proposed univariate linear model for the dataset $D=\left\{(x_1,\,y_1),\,(x_2,\,y_2),\,\ldots,\,(x_N,\,y_N)\right\}$, where $\theta_0,\,\theta_1$ are unknown parameters.

Then, the residuals are 

$$
\varepsilon_n=y_n-\hat{y}_n \tag*{(2)}
$$

and the error or loss function in this case is the mean squared error (MSE)

$$
\ell(\theta_0,\,\theta_1)=\frac{1}{N}\sum_{n=1}^N\varepsilon_n^2=\frac{1}{N}\sum_{n=1}^N(y_n-\hat{y}_n)^2=\frac{1}{N}\sum_{n=1}^N(y_n-\theta_0-\theta_1x_n)^2 \tag*{(3)}
$$

Minimizing the MSE yields closed-form expressions for the slope and intercept which are referred to as the ordinary least squares (OLS) estimates of the parameters $\theta_0$ and $\theta_1$:

$$
\hat{\theta}_0=\overline{y}-\hat{\theta}_1\overline{x},\qquad\hat{\theta}_1=\frac{\sum_{n=1}^N(x_n-\overline{x})(y_n-\overline{y})}{\sum_{n=1}^N(x_n-\overline{x})^2} \tag*{(4)}
$$

As a reminder, 

$$
\overline{x}=\frac{1}{N}\sum_{n=1}^Nx_n,\qquad\overline{y}=\frac{1}{N}\sum_{n=1}^Ny_n \tag*{(5)}
$$

## Multivariate Logistic Classification

Let $\sigma\colon\mathbb{R}\to\mathbb{R}$ be defined as

$$
\sigma(z)=\frac{1}{1+e^{-z}} \tag*{(6)}
$$

For a $k$-dimensional feature vector $\vec{x}=(x_1,\,x_2,\,\ldots,\,x_k)\in\mathbb{R}^{1\times k}$, the augmented feature vector is $\vec{x}^*=(1,\,x_1,\,x_2,\,\ldots,\,x_k)\in\mathbb{R}^{1\times(k+1)}$. 

Let the parameter vector satisfy $\vec{w}\in\mathbb{R}^{k+1}$ and

$$
z=\vec{x}^*\,\vec{w}=w_0+w_1x_1+w_2x_2+\cdots+w_kx_k \tag*{(7)}
$$

Hence, the multivariate logistic classifier predicts

$$
\hat{y}=\sigma(z) \tag*{(8)}
$$

where $\hat{y}\in(0,\,1)$ is interpreted as the probability of the positive class $(y=1)$, while the negative class is $y=0$.

The error or loss function is the binary cross-entropy (BCE):

$$
\ell(y,\,\hat{y})=-\big(y\ln(\hat{y})+(1-y)\ln(1-\hat{y})\big) \tag*{(9)}
$$

Since $\hat{y}=\sigma(z)$, differentiation with respect to $z$ gives

$$
\frac{d\ell}{dz}=\hat{y}-y \tag*{(10)}
$$

Therefore, by the chain rule, the gradient with respect to the parameter vector is

$$
\nabla\ell(\vec{w})=(\hat{y}-y)\,\vec{x}^{*\top} \tag*{(11)}
$$

Thus, the mean gradient over a dataset with $N$ observations is

$$
\nabla\mathscr{L}(\vec{w})=\frac{1}{N}\sum_{n=1}^N(\hat{y}_n-y_n)\,\vec{x}_n^{*\top}=\frac{1}{N}\sum_{n=1}^N\left(\sigma\left(\vec{x}_n^*\,\vec{w}\right)-y_n\right)\vec{x}_n^{*\top} \tag*{(12)}
$$

Unlike linear regression, there is no closed-form solution for the optimal parameter vector, so iterative optimization methods such as gradient descent are used to minimize the objective function $\mathscr{L}$ (not stated). 

Hence, the update rule is

$$
\vec{w}^{(t+1)}=\vec{w}^{(t)}-\xi\,\nabla\mathscr{L}\left(\vec{w}^{(t)}\right) \tag*{(13)}
$$

$$
=\vec{w}^{(t)}-\frac{\xi}{N}\sum_{n=1}^N\left(\sigma\left(\vec{x}_n^*\,\vec{w}^{(t)}\right)-y_n\right)\vec{x}_n^{*\top} \tag*{(14)}
$$

where $\xi\in\mathbb{R}^+$ is the learning rate and $t$ is the iteration of the parameter vector.
