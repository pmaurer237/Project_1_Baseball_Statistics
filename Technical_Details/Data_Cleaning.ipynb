## 1. Reviewing Baseball Stats

Baseball is a sport *deeply rooted in numbers*. For over a century, analysts, writers, and fans have used statistics to describe player performance, compare players across eras, and make in-game or front-office decisions. While traditional statistics such as batting average and home runs remain widely recognized, modern baseball analysis increasingly relies on more advanced metrics that paint a fuller picture of a player's true value. This overview introduces the most important statistics used in today's game, including many that are particularly relevant to this project.

---

### Traditional Batting Statistics

#### Plate Appearance (PA) and At-Bat (AB)

A PA is *every completed turn* a batter takes at the plate, regardless of outcome (hit, walk, strikeout, etc.). An AB is a type of PA that does not include walks (below), hit-by-pitch (HBP), and sacrifices. 

#### Batting Average (AVG)

AVG is one of the *oldest* stats in baseball and measures how often a batter gets a hit (H):

$$
\text{AVG}=\frac{\text{H}}{\text{AB}} \tag*{(1)}
$$

While easy to understand, AVG *ignores* walks, power, and quality of contact, which limits its usefulness.

#### Home Run (HR)

A HR occurs when a batter hits the ball *out of play* in fair territory or circles the bases on an in-play hit. HR totals are a simple but powerful indicator of a hitter's impact and especially relevant for power hitters like Aaron Judge.

#### Run Batted In (RBI)

An RBI occurs when a *baserunner scores* as the result of a batter's action, like a hit or walk. While popular, this depends heavily on teammates being on base and thus are not always ideal for evaluating individual skill.

#### Walk/Base-on-Balls (BB) and Strikeout (SO)

BBs measure *plate discipline* (a batter's ability to recognize balls outside the strike zone), while SOs are an indicator of both plate discipline and contact ability. Modern analytics highlight walk percentage and strikeout percentage (below), both expressed as percentages of plate appearances.

---

### Modern Batting Statistics

#### On-Base Percentage (OBP)

OBP improves on batting average by *including walks and hit-by-pitches*:

$$
\text{OBP}=\frac{\text{H}+\text{BB}+\text{HBP}}{\text{AB}+\text{BB}+\text{HBP}+\text{SF}} \tag*{(2)}
$$

A high OBP means a player *consistently reaches base*, which contributes directly to run scoring. Note that sacrifice fly (SF) count is included in the denominator.

#### Slugging Percentage (SLG)

SLG accounts for *how many bases* a player gains per at-bat:

$$
\text{SLG}=\frac{\text{Total Bases}}{\text{AB}} \tag*{(3)}
$$

*Extra-base hits* matter more here, making SLG a better measure of power than HR alone.

#### On-Base Plus Slugging (OPS)

OPS simply *adds* OBP and SLG. It's easy to compute and widely used to evaluate hitters' performances in a given season.

#### Isolated Power (ISO)

ISO isolates (hence the name) *extra-base* power:

$$
\text{ISO}=\text{SLG}-\text{AVG} \tag*{(4)}
$$

ISO is one of the cleanest ways to quantify *raw power*, an area where Judge excels annually.

---

### Plate Discipline Metrics (Basic)

Plate discipline is critical for distinguishing between *aggressive, selective, and patient* hitters. Key metrics include:

#### Walk Percentage (BB%)

Percentage of plate appearances that end in a BB. *Higher* is better.

#### Strikeout Percentage (SO%)

Percentage of plate appearances that end in a SO. *Lower* is better.

#### Walk-to-Strikeout Ratio (BB/SO)

A quick *composite* measure of overall plate discipline. Great hitters typically have $\text{BB}\%\geq\text{SO}\%$, but power hitters may strike out more while still producing excellent offensive value.

---

### Statcast Metrics (Contact Quality)

Statcast essentially measures the *physics* of batted balls. These modern statistics have transformed how analysts evaluate hitters.

#### Exit Velocity (EV)

The speed of the ball off the bat, measured in *miles per hour* (mph). Judge consistently leads MLB in both average EV and maximum EV.

#### Launch Angle (LA)

The *vertical angle* at which the ball leaves the bat. Ideal LA depends on hitting style.

#### Hard Hit Percentage

Percentage of batted balls hit at *$95$ mph or harder*. This is a key indicator of consistent quality contact.

#### Barrels

A "barrel" is a batted ball with the *optimal combination* of EV and LA (e.g., $98+$ mph at $26-30$ degrees). Barrel percentage is known to be one of the strongest predictors of overall power performance.

#### Sweet Spot Percentage

Percentage of batted balls hit *between $8$ degrees and $32$ degrees*. This is the range most conducive to line drives and elevated contact, often leading to HRs.

---

### Context-Adjusted Metrics

Some of the most important modern stats adjust performance for factors like *ballpark, era, and league run environment*.

#### Weighted On-Base Average (wOBA)

wOBA assigns *different weights* to singles (1B), doubles (2B), triples (3B), HR, and BB to approximate their true run value. It is a more accurate version of OPS.

#### Weighted Runs Created Plus (wRC+)

wRC+ improves wOBA by *adjusting for ballpark and run environment*:

- $100 =$ league average
- $150 = 50\%$ better than league average
- $200 = 100\%$ better

Aaron Judge *routinely* posts wRC+ values between $150–200$!

#### Wins Above Replacement (WAR)

WAR estimates how many wins a player adds above a "replacement-level" player. It incorporates *offense, base running, and defense*. This stat often determines who wins the most valuable player (MVP) awards at the end of a season.

---

### Summary

Modern baseball analysis blends traditional metrics (like AVG and OPS) with advanced sabermetrics (like wRC+ and ISO) and Statcast technology (like EV and barrel percentage). *Together*, these statistics enable a much deeper, more accurate understanding of a player's offensive performance. For a player like Aaron Judge, whose value is tied to both exceptional power and strong plate discipline, these metrics collectively help contextualize just how extraordinary his performance has been relative to his peers.

## 4. Column Name Standardization In `standard_raw_all` Dataset

```python
# Define a function to clean column names!
def clean_columns(df):
    df = df.copy()
    df.columns = (df.columns
                  .str.strip()
                  .str.lower()
                  .str.replace(" ", "_")
                  .str.replace("%", "pct")
                  .str.replace("/", "_")
                  .str.replace("-", "_")
                  .str.replace("(", "")
                  .str.replace(")", ""))
    return df
standard_raw_all = clean_columns(standard_raw_all)
standard_raw_judge = clean_columns(standard_raw_judge)
statcast_raw_all = clean_columns(statcast_raw_all)
statcast_raw_judge = clean_columns(statcast_raw_judge)

# Display cleaned column names for verification!
print("Example cleaned columns from FanGraphs data:\n")
print(standard_raw_all.columns[:25], "\n")
print("Example cleaned columns from Statcast data:\n")
print(statcast_raw_all.columns[:25])
```

## 8. Splitting Name Columns Into First And Last Names In Both Datasets

```python
# FanGraphs names!
def split_fangraphs_name(full_name):
    if pd.isna(full_name):
        return np.nan, np.nan
    parts = full_name.strip().split()
    if len(parts) == 1:
        return parts[0], np.nan # Very rare case!
    return parts[0], " ".join(parts[1:])

# Apply split!
standard_clean_all["first_name"] = standard_clean_all["name"].apply(lambda x: split_fangraphs_name(x)[0])
standard_clean_all["last_name"] = standard_clean_all["name"].apply(lambda x: split_fangraphs_name(x)[1])

# Statcast names!
def split_statcast_name(full_name):
    if pd.isna(full_name):
        return np.nan, np.nan
    if "," in full_name:
        last, first = full_name.split(",", 1)
        return first.strip(), last.strip()
    
    # Fallback: If Statcast ever gives "Aaron Judge"!
    parts = full_name.strip().split()
    if len(parts) == 1:
        return parts[0], np.nan
    return parts[0], " ".join(parts[1:])

# Apply split!
statcast_clean_all["first_name"] = statcast_clean_all["player_name"].apply(lambda x: split_statcast_name(x)[0])
statcast_clean_all["last_name"] = statcast_clean_all["player_name"].apply(lambda x: split_statcast_name(x)[1])

# Drop original name columns!
standard_clean_all = standard_clean_all.drop(columns = ["name"])
statcast_clean_all = statcast_clean_all.drop(columns = ["player_name"])
```

## 9. Standardizing Datatypes In `standard_raw_all` Dataset

```python
# Make a copy!
standard_clean_all = standard_clean_all.copy()

# Columns that must remain non-numeric!
non_numeric_cols = ["team", "first_name", "last_name"]

# Identify numeric columns explicitly!
numeric_cols = [col for col in standard_clean_all.columns if col not in non_numeric_cols]

# Convert all numeric-like columns to numeric dtype!
standard_clean_all[numeric_cols] = standard_clean_all[numeric_cols].apply(pd.to_numeric, errors = "coerce")

# Ensure ID-like columns are nullable integers!
id_like_cols = ["fangraphs_id", "mlbam_id", "season", "age", "games", "ab", "pa"]
for col in id_like_cols:
    if col in standard_clean_all.columns:
        standard_clean_all[col] = standard_clean_all[col].astype("Int64")
```

## 10. Merging The Two Cleaner Datasets

```python
# Make copies to avoid accidental modification!
fg = standard_clean_all.copy()
sc = statcast_clean_all.copy()

# Ensure numeric IDs for merge!
fg["mlbam_id"] = fg["mlbam_id"].astype("Int64")
sc["mlbam_id"] = sc["mlbam_id"].astype("Int64")
fg["season"] = fg["season"].astype("Int64")
sc["season"] = sc["season"].astype("Int64")

# Perform left merge; keep all FanGraphs rows!
combo_clean_all = fg.merge(sc,
                           on = ["mlbam_id", "season"],
                           how = "left",
                           suffixes = ("", "_statcast"))
```

## 11. Dropping Some More Unnecessary Columns From The Merged Dataset And Reordering Columns

```python
# Drop columns!
combo_clean_all = combo_clean_all.drop(columns = {"ultimate_base_running", 
                                                  "first_name_statcast", 
                                                  "last_name_statcast", 
                                                  "events",
                                                  "off",
                                                  "def",
                                                  "speed_score",
                                                  "base_running",
                                                  "age",
                                                  "max_exit_velocity_statcast",
                                                  "barrels_statcast",
                                                  "barrel_percentage_statcast",
                                                  "barrels_per_pa_statcast",
                                                  "ibb",
                                                  "average_exit_velocity",
                                                  "average_launch_angle",
                                                  "exit_velocity_95th_percentile",
                                                  "avg_exit_velocity_50th_percentile"})

# Rename columns!
combo_clean_all = combo_clean_all.rename(columns = {"war": "wins_above_replacement",
                                                    "barrelpct": "barrel_percentage",
                                                    "hardhitpct": "hard_hit_percentage",
                                                    "exit_velocity": "avg_exit_velocity",
                                                    "avg": "batting_avg",
                                                    "batting_average_on_balls_in_play": "batting_avg_on_balls_in_play",
                                                    "bbpct": "bb_percentage",
                                                    "kpct": "so_percentage",
                                                    "swingpct": "swing_percentage",
                                                    "contactpct": "contact_percentage",
                                                    "zonepct": "zone_percentage",
                                                    "pullpct": "pull_percentage",
                                                    "centpct": "cent_percentage",
                                                    "oppopct": "oppo_percentage",
                                                    "hardhit": "hard_hit_balls",
                                                    "flyball_linedrive_exit_velocity": "fly_ball_line_drive_exit_velocity",
                                                    "groundball_exit_velocity": "ground_ball_exit_velocity",
                                                    "launch_angle": "avg_launch_angle",
                                                    "exit_velocity_95_mph_plus": "exit_velocity_95_mph_plus_count",
                                                    "average_distance_hit": "avg_distance_hit"})

# Deal with duplicated columns!
combo_clean_all = combo_clean_all.loc[:, ~combo_clean_all.columns.duplicated(keep = "last")].copy()

# Reorder columns!
combo_clean_all = combo_clean_all.copy()
first_cols = ["fangraphs_id", "mlbam_id", "first_name", "last_name", "team", "season"]
remaining_cols = [col for col in combo_clean_all.columns if col not in first_cols]
new_col_order = first_cols + remaining_cols
combo_clean_all = combo_clean_all[new_col_order].round(3)

# Preview!
print("Preview of clean dataset:\n")
combo_clean_all.head()
```

```python
# Judge-only dataset!
combo_clean_judge = combo_clean_all[combo_clean_all["last_name"] == "Judge"].copy()

# Export both datasets!
combo_clean_all.to_csv("../data/clean/combo_clean_all.csv", index = False)
combo_clean_judge.to_csv("../data/clean/combo_clean_judge.csv", index = False)
```
