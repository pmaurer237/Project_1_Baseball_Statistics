## Column Name Standardization 

```python
# Define a function to standardize column names
def standardize_columns(df):
    df = df.copy()
    df.columns = (df.columns
                  .str.strip()
                  .str.lower()
                  .str.replace(" ", "_")
                  .str.replace("%", "pct")
                  .str.replace("/", "_")
                  .str.replace("-", "_")
                  .str.replace("(", "")
                  .str.replace(")", ""))
    return df

# Call the standardization function on each dataset
fangraphs_raw = standardize_columns(fangraphs_raw)
statcast_raw = standardize_columns(statcast_raw)

# Display some column names for verification
print("Some clean column names from the FanGraphs dataset:\n")
print(fangraphs_raw.columns[:10], "\n\n")
print("Some clean column names from the Statcast dataset:\n")
print(statcast_raw.columns[:10])
```

## 8. Splitting Name Columns Into First And Last Names In Both Datasets

```python
# FanGraphs names!
def split_fangraphs_name(full_name):
    if pd.isna(full_name):
        return np.nan, np.nan
    parts = full_name.strip().split()
    if len(parts) == 1:
        return parts[0], np.nan # Very rare case!
    return parts[0], " ".join(parts[1:])

# Apply split!
standard_clean_all["first_name"] = standard_clean_all["name"].apply(lambda x: split_fangraphs_name(x)[0])
standard_clean_all["last_name"] = standard_clean_all["name"].apply(lambda x: split_fangraphs_name(x)[1])

# Statcast names!
def split_statcast_name(full_name):
    if pd.isna(full_name):
        return np.nan, np.nan
    if "," in full_name:
        last, first = full_name.split(",", 1)
        return first.strip(), last.strip()
    
    # Fallback: If Statcast ever gives "Aaron Judge"!
    parts = full_name.strip().split()
    if len(parts) == 1:
        return parts[0], np.nan
    return parts[0], " ".join(parts[1:])

# Apply split!
statcast_clean_all["first_name"] = statcast_clean_all["player_name"].apply(lambda x: split_statcast_name(x)[0])
statcast_clean_all["last_name"] = statcast_clean_all["player_name"].apply(lambda x: split_statcast_name(x)[1])

# Drop original name columns!
standard_clean_all = standard_clean_all.drop(columns = ["name"])
statcast_clean_all = statcast_clean_all.drop(columns = ["player_name"])
```

## 9. Standardizing Datatypes In `standard_raw_all` Dataset

```python
# Make a copy!
standard_clean_all = standard_clean_all.copy()

# Columns that must remain non-numeric!
non_numeric_cols = ["team", "first_name", "last_name"]

# Identify numeric columns explicitly!
numeric_cols = [col for col in standard_clean_all.columns if col not in non_numeric_cols]

# Convert all numeric-like columns to numeric dtype!
standard_clean_all[numeric_cols] = standard_clean_all[numeric_cols].apply(pd.to_numeric, errors = "coerce")

# Ensure ID-like columns are nullable integers!
id_like_cols = ["fangraphs_id", "mlbam_id", "season", "age", "games", "ab", "pa"]
for col in id_like_cols:
    if col in standard_clean_all.columns:
        standard_clean_all[col] = standard_clean_all[col].astype("Int64")
```

## 10. Merging The Two Cleaner Datasets

```python
# Make copies to avoid accidental modification!
fg = standard_clean_all.copy()
sc = statcast_clean_all.copy()

# Ensure numeric IDs for merge!
fg["mlbam_id"] = fg["mlbam_id"].astype("Int64")
sc["mlbam_id"] = sc["mlbam_id"].astype("Int64")
fg["season"] = fg["season"].astype("Int64")
sc["season"] = sc["season"].astype("Int64")

# Perform left merge; keep all FanGraphs rows!
combo_clean_all = fg.merge(sc,
                           on = ["mlbam_id", "season"],
                           how = "left",
                           suffixes = ("", "_statcast"))
```

## 11. Dropping Some More Unnecessary Columns From The Merged Dataset And Reordering Columns

```python
# Drop columns!
combo_clean_all = combo_clean_all.drop(columns = {"ultimate_base_running", 
                                                  "first_name_statcast", 
                                                  "last_name_statcast", 
                                                  "events",
                                                  "off",
                                                  "def",
                                                  "speed_score",
                                                  "base_running",
                                                  "age",
                                                  "max_exit_velocity_statcast",
                                                  "barrels_statcast",
                                                  "barrel_percentage_statcast",
                                                  "barrels_per_pa_statcast",
                                                  "ibb",
                                                  "average_exit_velocity",
                                                  "average_launch_angle",
                                                  "exit_velocity_95th_percentile",
                                                  "avg_exit_velocity_50th_percentile"})

# Rename columns!
combo_clean_all = combo_clean_all.rename(columns = {"war": "wins_above_replacement",
                                                    "barrelpct": "barrel_percentage",
                                                    "hardhitpct": "hard_hit_percentage",
                                                    "exit_velocity": "avg_exit_velocity",
                                                    "avg": "batting_avg",
                                                    "batting_average_on_balls_in_play": "batting_avg_on_balls_in_play",
                                                    "bbpct": "bb_percentage",
                                                    "kpct": "so_percentage",
                                                    "swingpct": "swing_percentage",
                                                    "contactpct": "contact_percentage",
                                                    "zonepct": "zone_percentage",
                                                    "pullpct": "pull_percentage",
                                                    "centpct": "cent_percentage",
                                                    "oppopct": "oppo_percentage",
                                                    "hardhit": "hard_hit_balls",
                                                    "flyball_linedrive_exit_velocity": "fly_ball_line_drive_exit_velocity",
                                                    "groundball_exit_velocity": "ground_ball_exit_velocity",
                                                    "launch_angle": "avg_launch_angle",
                                                    "exit_velocity_95_mph_plus": "exit_velocity_95_mph_plus_count",
                                                    "average_distance_hit": "avg_distance_hit"})

# Deal with duplicated columns!
combo_clean_all = combo_clean_all.loc[:, ~combo_clean_all.columns.duplicated(keep = "last")].copy()

# Reorder columns!
combo_clean_all = combo_clean_all.copy()
first_cols = ["fangraphs_id", "mlbam_id", "first_name", "last_name", "team", "season"]
remaining_cols = [col for col in combo_clean_all.columns if col not in first_cols]
new_col_order = first_cols + remaining_cols
combo_clean_all = combo_clean_all[new_col_order].round(3)

# Preview!
print("Preview of clean dataset:\n")
combo_clean_all.head()
```
