# Data Cleaning

## Necessary Packages

```python
import pandas as pd
import numpy as np
```

## Dataset Load

```python
# Load raw datasets
fangraphs_raw = pd.read_csv("../Data/Raw/FanGraphs_Raw.csv")
statcast_raw = pd.read_csv("../Data/Raw/Statcast_Raw.csv")
```

## Column Name Formatting

```python
# Define a function to standardize column names
def standardize_columns(df):
    df.columns = (df.columns
                  .str.strip()
                  .str.lower()
                  .str.replace(" ", "_")
                  .str.replace("%", "pct")
                  .str.replace("/", "_")
                  .str.replace("-", "_")
                  .str.replace("(", "")
                  .str.replace(")", ""))
    return df

# Call the standardization function on each dataset
fangraphs_raw = standardize_columns(fangraphs_raw)
statcast_raw = standardize_columns(statcast_raw)

# Display some column names for verification
print("Some clean column names from the FanGraphs dataset:\n")
print(fangraphs_raw.columns[:25], "\n\n")
print("Some clean column names from the Statcast dataset:\n")
print(statcast_raw.columns[:25])
```

## Smaller Version of FanGraphs Dataset with Only Relevant Columns

```python
# Core identifiers
id_cols = ["fangraphs_id", 
           "mlbam_id", 
           "season", 
           "player_name", 
           "team"]

# Counting stats
counting = ["games_played", 
            "at_bats", 
            "plate_appearances", 
            "hits", 
            "singles", 
            "doubles", 
            "triples", 
            "home_runs", 
            "runs", 
            "runs_batted_in", 
            "walks", 
            "intentional_walks", 
            "strikeouts", 
            "hit_by_pitch", 
            "stolen_bases", 
            "caught_stealing"]

# Traditional rate stats
traditional_rate = ["batting_average", 
                    "on_base_percentage", 
                    "slugging_percentage", 
                    "on_base_plus_slugging", 
                    "isolated_power", 
                    "batting_average_on_balls_in_play"]

# Plate discipline
discipline = ["walk_percentage", 
              "strikeout_percentage", 
              "walks_to_strikeouts_ratio",
              "outside_swing_percentage", 
              "zone_swing_percentage", 
              "swing_percentage",
              "outside_contact_percentage", 
              "zone_contact_percentage", 
              "contact_percentage", 
              "zone_percentage", 
              "first_pitch_strike_percentage", 
              "swinging_strike_percentage"]

# Batted-ball profile
batted_ball = ["ground_balls", 
               "fly_balls", 
               "line_drives",
               "infield_fly_balls",
               "ground_ball_percentage", 
               "line_drive_percentage",
               "fly_ball_percentage", 
               "infield_fly_ball_percentage",
               "home_run_fly_ball_percentage", 
               "pull_percentage", 
               "center_percentage", 
               "oppo_percentage", 
               "soft_hit_percentage", 
               "medium_hit_percentage", 
               "hard_hit_percentage"]

# Value stats
value_stats = ["weighted_on_base_average", 
               "weighted_runs_above_average",
               "weighted_runs_created", 
               "weighted_runs_created_plus",
               "runs_above_replacement", 
               "wins_above_replacement", 
               "speed_score", 
               "base_running_score", 
               "ultimate_base_running_score"]

# Statcast-related columns
statcast_related = ["average_exit_velocity", 
                    "average_launch_angle", 
                    "barrels", 
                    "barrel_percentage",
                    "maximum_exit_velocity", 
                    "called_strike_percentage", 
                    "called_strike_whiff_percentage", 
                    "expected_batting_average", 
                    "expected_slugging_percentage", 
                    "expected_weighted_on_base_average"]

# Create the final keep list
fangraphs_keep_cols = (id_cols + counting + traditional_rate +
                       discipline + batted_ball + value_stats +
                       statcast_related)

# Create the smaller dataset
fangraphs_raw = fangraphs_raw[fangraphs_keep_cols]
print(f"Shape of the smaller FanGraphs dataset: {fangraphs_raw.shape}\n")
fangraphs_raw.head()
```

## Names of Relevant Columns in Statcast Dataset

```python
# Again, these relevant columns must be renamed manually
statcast_raw = statcast_raw.rename(
    columns = {
        "player_id": "mlbam_id",  # This is a duplicate but will be kept for merging
        "anglesweetspotpercent": "sweet_spot_percentage", 
        "fbld": "fly_ball_line_drive_average_exit_velocity",
        "gb": "ground_ball_average_exit_velocity",
        "max_distance": "maximum_distance_hit",
        "avg_distance": "average_distance_hit",
        "avg_hr_distance": "average_home_run_distance",
        "ev95plus": "exit_velocity_95_mph_plus_count",
        "brl_pa": "barrels_per_plate_appearance_percentage",
               }
                                    )
```

## Smaller Version of Statcast Dataset with Only Relevant Columns

```python
# Create the final keep list
statcast_keep_cols = ["mlbam_id", 
                      "season",  # This is a duplicate but will be kept for merging 
                      "sweet_spot_percentage",
                      "fly_ball_line_drive_average_exit_velocity", 
                      "ground_ball_average_exit_velocity",
                      "maximum_distance_hit", 
                      "average_distance_hit", 
                      "average_home_run_distance", 
                      "exit_velocity_95_mph_plus_count", 
                      "barrels_per_plate_appearance_percentage"]

# Create the smaller dataset
statcast_raw = statcast_raw[statcast_keep_cols]
print(f"Shape of the smaller Statcast dataset: {statcast_raw.shape}\n")
statcast_raw.head()
```

## FanGraphs and Statcast Merge

```python
# Merge on shared keys
fangraphs_statcast_raw = fangraphs_raw.merge(statcast_raw,
                                             how = "left",
                                             on = ["mlbam_id", "season"])
print(f"Shape of the merged dataset: {fangraphs_statcast_raw.shape}\n")
fangraphs_statcast_raw.head()
```

## Missing Values

```python
# Drop rows with missing values
fangraphs_statcast_raw = fangraphs_statcast_raw.dropna()
print(f"Shape of the merged dataset after dropping missing values: {fangraphs_statcast_raw.shape}\n")
fangraphs_statcast_raw.head()
```

## Datatypes

```python
# Identify non-numeric columns 
non_numeric_cols = ["player_name", "team"]

# Convert everything else to numeric
for col in fangraphs_statcast_raw.columns:
    if col not in non_numeric_cols:
        fangraphs_statcast_raw[col] = pd.to_numeric(fangraphs_statcast_raw[col], errors = "coerce")

# Round all numeric columns to at most three decimal places
numeric_cols = fangraphs_statcast_raw.select_dtypes(include = ["number"]).columns
fangraphs_statcast_raw[numeric_cols] = fangraphs_statcast_raw[numeric_cols].round(3)

# Some columns require special handling (not in order below)
fangraphs_statcast_raw["sweet_spot_percentage"] = (fangraphs_statcast_raw["sweet_spot_percentage"] / 100).round(3)
fangraphs_statcast_raw["barrels_per_plate_appearance_percentage"] = (fangraphs_statcast_raw["barrels_per_plate_appearance_percentage"] / 100).round(3)
fangraphs_statcast_raw["exit_velocity_95_mph_plus_count"] = fangraphs_statcast_raw["exit_velocity_95_mph_plus_count"].astype(int)
fangraphs_statcast_raw["barrels"] = fangraphs_statcast_raw["barrels"].astype(int)
fangraphs_statcast_raw["maximum_distance_hit"] = fangraphs_statcast_raw["maximum_distance_hit"].astype(int)
fangraphs_statcast_raw["average_distance_hit"] = fangraphs_statcast_raw["average_distance_hit"].astype(int)
fangraphs_statcast_raw["average_home_run_distance"] = fangraphs_statcast_raw["average_home_run_distance"].astype(int)
fangraphs_statcast_raw["weighted_runs_created_plus"] = fangraphs_statcast_raw["weighted_runs_created_plus"].astype(int)

# Check result
fangraphs_statcast_raw.head()
```

## Final Tasks

```python
# Remove ID columns that are no longer needed
final_cols = [col for col in fangraphs_statcast_raw.columns if col not in ["fangraphs_id", 
                                                                           "mlbam_id"]]
clean_dataset = fangraphs_statcast_raw[final_cols]

# Make a Judge-only dataset
judge_dataset = clean_dataset[clean_dataset["player_name"] == "Aaron Judge"].copy()

# Save the clean datasets
clean_dataset.to_csv("../Data/Clean/All_Players.csv", index = False)
judge_dataset.to_csv("../Data/Clean/Judge_Only.csv", index = False)
```
