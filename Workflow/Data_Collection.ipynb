## FanGraphs (Standard and Advanced Batting) Stats Collection (end of demonstration)

### Setup

```python
# Import necessary packages
import pandas as pd 

# Set some constants
START_SEASON = 2017
END_SEASON = 2025
MIN_PA = 200  # Set the minimum number of season plate appearances to include a player
SEASONS = list(range(START_SEASON, END_SEASON + 1))
```

### Pull (2017-2025)

```python
# Pull FanGraphs stats for all MLB hitters from 2017 to 2025, excluding 2020
fangraphs_list = []
for season in SEASONS:
    if season == 2020:
        continue  # Skip COVID season
    df_f = batting_stats(season, qual = 0)  
    
    # Filter for hitters with at least 200 plate appearances in the given season
    df_f = df_f[df_f["PA"] >= MIN_PA].copy()
    fangraphs_list.append(df_f)

# Concatenate all seasons 
fg_raw_all = pd.concat(fangraphs_list, ignore_index = True)
print(f"Shape of FanGraphs dataset for all players initially: {fg_raw_all.shape}\n")
fg_raw_all.head()
```

### FanGraphs IDs

```python
# Fix the FanGraphs ID column name 
fg_raw_all = fg_raw_all.rename(columns = {"IDfg": "fangraphs_id"})

# Ensure that FanGraphs IDs are integers
fg_raw_all["fangraphs_id"] = pd.to_numeric(fg_raw_all["fangraphs_id"], errors = "coerce").astype("Int64")

# Check for missing FanGraphs IDs
missing_fg = fg_raw_all["fangraphs_id"].isna().sum()
print(f"Missing FanGraphs IDs: {missing_fg}\n")

# Display duplicates by name and season
duplicates = fg_raw_all[fg_raw_all.duplicated(subset = ["Name", "Season"], keep = False)]
print(f"Number of duplicate name-season pairs: {duplicates.shape[0]}\n")
duplicates.head()
```

Note: There are currently two *different* players named `"Max Muncy"` in the MLB!

## Now Statcast Data Collection!

### ID Map: FanGraphs $\longrightarrow$ MLBAM (for Statcast)

```python
from pybaseball import playerid_reverse_lookup
```

```python
# Collect unique FanGraphs IDs from the FanGraphs dataset
unique_fangraphs_ids = fg_raw_all["fangraphs_id"].dropna().astype(int).unique().tolist()

# Look up cross-IDs for all of these FanGraphs IDs
id_map = playerid_reverse_lookup(unique_fangraphs_ids, key_type = "fangraphs")

# Keep the relevant IDs and rename the columns
id_map = id_map.rename(columns = {"key_fangraphs": "fangraphs_id",
                                  "key_mlbam": "mlbam_id"})
id_map = id_map[["fangraphs_id", "mlbam_id"]]
print("IDs:\n")
id_map.head()
```

### Merge New MLBAM ID Column with `fg_raw_all`

```python
# Left-join the MLBAM ID column to the FanGraphs dataset
fg_raw_all = fg_raw_all.merge(id_map, on = "fangraphs_id", how = "left")

# Ensure that MLBAM IDs are integers
fg_raw_all["mlbam_id"] = pd.to_numeric(fg_raw_all["mlbam_id"], errors = "coerce").astype("Int64")

# Check for missing MLBAM IDs
missing_mlbam = fg_raw_all["mlbam_id"].isna().sum()
print(f"Missing MLBAM IDs: {missing_mlbam}\n")
fg_raw_all.head()
```

### Remove Players with no MLBAM ID

```python
# Remove players with no MLBAM ID since they have no Statcast data
before = fg_raw_all.shape[0]
fg_raw_all = fg_raw_all.dropna(subset = ["mlbam_id"]).copy()
fg_raw_all["mlbam_id"] = fg_raw_all["mlbam_id"].astype(int)
after = fg_raw_all.shape[0]

# Report how many players were removed
print(f"Removed {before - after} players with missing MLBAM IDs!\n")
print(f"Remaining rows: {after}\n")
fg_raw_all.head()
```

### NOW Save FanGraphs Datasets

```python
# Make a separate FanGraphs dataset for just Aaron Judge 
judge_fangraphs_id = 15640
fg_raw_judge = fg_raw_all[fg_raw_all["fangraphs_id"] == judge_fangraphs_id].copy()
print(f"Shape of FanGraphs dataset for Judge only: {fg_raw_judge.shape}")

# Save the datasets
fg_raw_all.to_csv("../Data/Raw/FanGraphs_Raw_All_Players.csv", index = False)
fg_raw_judge.to_csv("../Data/Raw/FanGraphs_Raw_Judge_Only.csv", index = False)
```

### Statcast Pull (2017-2025)

```python
from pybaseball import statcast_batter_exitvelo_barrels
```

```python
# Pull Statcast aggregates for all hitters from 2017 to 2025, excluding 2020
statcast_list = []
for season in SEASONS:
    if season == 2020:  
        continue  # Skip COVID season
    df_s = statcast_batter_exitvelo_barrels(season)
    df_s["Season"] = season
    statcast_list.append(df_s)

# Concatenate all seasons 
statcast_raw_all = pd.concat(statcast_list, ignore_index = True)
print(f"Shape of Statcast dataset for all players initially: {statcast_raw_all.shape}\n")
statcast_raw_all.head()
```

Note: The Statcast batting dataset is *smaller (in terms of rows)* than the FanGraphs batting dataset because Statcast only includes players who generated measurable batted-ball events in a given season. While FanGraphs reports statistics for every hitter who recorded plate appearances, Statcast requires actual tracked contact, such as exit velocity or launch angle, to generate a profile. As a result, players with *limited playing time, few balls in play, injury-shortened seasons, or data inconsistencies* may be present in FanGraphs but missing from Statcast. This discrepancy is expected and does not affect the analysis, since all meaningful comparison hitters, including Aaron Judge, have complete Statcast data.

### Remove Rows, Save Statcast Datasets

```python
# Keep Statcast rows only for hitters who appear in the FanGraphs dataset
valid_mlbam_ids = fg_raw_all["mlbam_id"].unique().tolist()
statcast_raw_all = statcast_raw_all[statcast_raw_all["player_id"].isin(valid_mlbam_ids)].copy()
print(f"Statcast rows after filtering to hitters with FanGraphs data: {statcast_raw_all.shape[0]}\n")

# Make a separate Statcast dataset for just Aaron Judge 
judge_mlbam_id = 592450
statcast_raw_judge = statcast_raw_all[statcast_raw_all["player_id"] == judge_mlbam_id].copy()
print(f"Shape of Statcast dataset for Judge only: {statcast_raw_judge.shape}")

# Save the datasets
statcast_raw_all.to_csv("../Data/Raw/Statcast_Raw_All_Players.csv", index = False)
statcast_raw_judge.to_csv("../Data/Raw/Statcast_Raw_Judge_Only.csv", index = False)
```
